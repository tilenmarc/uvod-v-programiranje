```python
zanri = pd.read_csv("zanri.csv")
```

```python
knjige_zanri = pd.merge(knjige, zanri, left_on="id", right_on="id_knjige")
```

```python
knjige_zanri.groupby("zanr")["trenutno_bere"].mean().sort_values(ascending=False).head(30).plot.bar()
```

```python
fan = knjige_zanri[(knjige_zanri["zanr"] == "Fantasy")]
rom = knjige_zanri[(knjige_zanri["zanr"] == "Romance")]
pd.merge(fan, rom[["id_knjige", "zanr"]], right_on="id_knjige", left_on="id_knjige").sort_values("trenutno_bere", ascending=False)[["naslov", "trenutno_bere"]]
```

```python
top_zanri = (
    knjige_zanri.groupby("zanr")[["trenutno_bere", "povprecna_ocena"]]
    .mean()
    .sort_values("trenutno_bere", ascending=False)
    .head(30)
)

ax = top_zanri["trenutno_bere"].plot.bar(figsize=(16, 6), color="skyblue")
top_zanri["povprecna_ocena"].plot(ax=ax, secondary_y=True, color="red", marker="o")

ax.set_ylabel("Povprecno trenutno bere")
ax.right_ax.set_ylabel("Povprecna ocena")
_ = ax.set_xticklabels(ax.get_xticklabels(), rotation=45, ha="right")
```

```python
dat = open("bunny.off")
dat.readline()
vrstica = dat.readline()
st_tock, st_ploskev = int(vrstica.split(" ")[0]), int(vrstica.split(" ")[1]) 

tocke = np.loadtxt(dat, max_rows=st_tock)
ploskve = np.loadtxt(dat, max_rows=st_ploskev)
```

```python
tocke
```

```python
ploskve =  ploskve.astype("int")
```

```python
trik = tocke[ploskve[:,1:]]
```

```python
vec1 = trik[:,1,:] - trik[:,0,:]
vec2 = trik[:,2,:] - trik[:,0,:]
```

```python
ploscina_zajca = np.sum(np.linalg.norm(np.cross(vec1, vec2, axis=1), axis=1)) / 2
ploscina_zajca
```
